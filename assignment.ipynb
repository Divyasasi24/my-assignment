```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, MinMaxScaler, RobustScaler

data = {
    'Gender': ['Male', 'Female', 'Female', 'Male', 'Female'],
    'City': ['Mumbai', 'Delhi', 'Chennai', 'Mumbai', 'Delhi'],
    'Size': ['Small', 'Large', 'Medium', 'Large', 'Small'],
    'Age': [25, 45, 32, 60, 28],
    'Fare': [15, 300, 85, 450, 20]
}

df = pd.DataFrame(data)

# ── Task 1: Categorical Encoding ──────────────────────────────────────────────

# 1a. Gender → Label Encoding (binary: Male/Female maps naturally to 0/1)
le_gender = LabelEncoder()
df['Gender'] = le_gender.fit_transform(df['Gender'])
# Female=0, Male=1  (alphabetical order by default)

# 1b. City → One-Hot Encoding, drop_first=True to avoid dummy variable trap
df = pd.get_dummies(df, columns=['City'], drop_first=True)
# Drops 'City_Chennai' as the reference category; adds City_Delhi, City_Mumbai

# 1c. Size → Ordinal Encoding with explicit order: Small=0, Medium=1, Large=2
size_order = {'Small': 0, 'Medium': 1, 'Large': 2}
df['Size'] = df['Size'].map(size_order)

# ── Task 2: Feature Scaling ───────────────────────────────────────────────────

# Fare has an outlier (450 vs typical 15–85), so RobustScaler is used for Fare.
# RobustScaler uses the median and IQR, making it resistant to outliers.
# Age has no extreme outliers; MinMaxScaler normalises it cleanly to [0, 1].

scaler_age  = MinMaxScaler()
scaler_fare = RobustScaler()

df['Age_scaled']  = scaler_age.fit_transform(df[['Age']])
df['Fare_scaled'] = scaler_fare.fit_transform(df[['Fare']])

df.drop(columns=['Age', 'Fare'], inplace=True)

print(df.to_string(index=False))
```
